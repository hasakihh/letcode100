```python
#查询key是否存在于字典的方法：dic.get(key)和key in dic
dic = {1:'one', 2: 'two', 3: 'three', 4: 'four', 5: 'five'}
if 3 in dic:
    print("Key 3 is present in the dictionary.")
if dic.get(6) is None:
    print("Key 6 is not present in the dictionary.")

#查询value是否存在于字典的方法：value in dic.values()
if 'three' in dic.values():
    print("Value 'three' is present in the dictionary.")

#查询字典中所有的key和value的方法：dic.keys()和dic.values()
print("All keys in the dictionary:", list(dic.keys()))
print("All values in the dictionary:", list(dic.values()))

#查询字典中所有的key-value对的方法：dic.items()
print("All key-value pairs in the dictionary:", list(dic.items())) #以列表的形式返回字典中所有的key-value对，列表内的每个元素都是一个元组
```

```python
#解1
#核心：1.发现排序后的元素都是一样的 2.在字典里创建一个空列表，这样就可以一个key对应多个”value“了
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        dic = {} #创建一个空字典
        for s in strs:#遍历列表里每一个字符串
            s_sorted = ''.join(sorted(s))   #！逐一对元素进行排序,获得排好序的元素了
            if s_sorted not in dic:         #
                dic[s_sorted] = []          #！创建一个空列表，这样就可以一个key对应多个”value“了
            dic[s_sorted].append(s)
        return list(dic.values())
    
## 做题经验：哈希表的value是题目要求最后返回那一类
```
