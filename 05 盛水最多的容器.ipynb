给定一个长度为 n 的整数数组 height 。有 n 条垂线，第 i 条线的两个端点是 (i, 0) 和 (i, height[i]) 。

找出其中的两条线，使得它们与 x 轴共同构成的容器可以容纳最多的水。

返回容器可以储存的最大水量。

说明：你不能倾斜容器。



输入：[1,8,6,2,5,4,8,3,7]
输出：49 
解释：图中垂直线代表输入数组 [1,8,6,2,5,4,8,3,7]。在此情况下，容器能够容纳水（表示为蓝色部分）的最大值为 49。

```python
#双指针：
#，面积=底*高，底是两个指针之间的距离，高是两个指针中较小的那个值。
#核心：
# 移动短板：那么面积可能增大，因为虽然底会变小，但高会增大
# 移动长板：面积一定减小，高不变，底变小，面积一定减小
#利用这个特性，只需要移动短板的指针就可以+1或者-1，直到两个指针相遇（left == right）。
class Solution:
    def maxArea(self, height: List[int]) -> int:
        left = 0
        right = len(height) - 1
        result = 0
        while left < right:
            if height[left] < height[right]:
                result = max(result, height[left] * (right - left))
                left += 1
            else:
                result = max(result, height[right] * (right - left))
                right -= 1
        return result
```

```python
#暴力解法：
# 两个for循环，遍历所有情况
class Solution:
    def maxArea(self, height: List[int]) -> int:
        S = 0
        n = len(height)
        for i in range(n-1):
            for j in range(i+1,n):
                s = (j-i)*min(height[i],height[j])
                S = max(s,S)
        return S
```
