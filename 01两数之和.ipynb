```python
#优化解
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        dic = {}
        # 收集nums下标及元素
        for i,j in enumerate(nums): #j = nums[i]
            val = target - j
            if val in dic:
                return [i,dic[val]]
            dic[j] = i
```

```python
# 解法1：字典哈希
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        
        dic = {}
        
        for i in range(len(nums)):
            val = target - nums[i]

            if dic.get(val) != None:
                return [dic[val],i]

            dic[nums[i]] = i

        return []

```

```python
#解法2 直接在列表里检索目标值，但是太麻烦，因为创建了很多切片，且新切片的下表会变
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:        
        for i in range(len(nums)):
            val = target - nums[i]
            if val in nums[i+1:len(nums)]:
                idx = nums[i+1:len(nums)].index(val)+1+i
                return [i,idx]

```
