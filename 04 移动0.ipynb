```python
#最优解
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        #解法3：双指针：(快慢双指针+交换法)
        #一个快指针，去遍历所有的非0元素
        #一个满指针，去标注非0元素该存放的位置
        slow = 0    #慢指针，第一个非0元素应该放在nums[0]
        for fast in range(len(nums)):   #快指针，for循环遍历找非0元素：
            if nums[fast] != 0:        #如果是非0元素，则交换快慢指针的位置
                nums[slow],nums[fast] = nums[fast],nums[slow] #一定要连着写，Python 会先计算右边所有的值，然后赋给左边，分开的话，会混乱

                slow += 1
    #如果都是0元素，则不交换，慢指针不动，快指针继续往后找非0元素
    #如果都是非0元素，则交换了两次（等于没换），慢指针和快指针都往后移动1位
```

```python

# 解法1：暴力
        # 时间复杂度为O(n)^2，因为每次.count和.remove都需要消耗O(n)
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        rm = 0
        while nums.count(0):
            nums.remove(0)
            rm += 1
        nums.extend([0]*rm)

#extend()--逐一添加元素
# append()--整体添加一个元素
# nums = [1 ,3 ,0 ,0 ,2 , 2, 4, 6, 5]
# zero_nums = nums.count(0)
# nums.append([0]* zero_nums) # 整体添加：[1, 3, 0, 0, 2, 2, 4, 6, 5, [0, 0]]
# nums.extend([0]* zero_nums)  # 逐一添加：[1, 3, 0, 0, 2, 2, 4, 6, 5, 0, 0]
```

```python
        
        #解法2：栈
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        # 1.初始化一个新栈
        stack = 0
        # 2. 遍历数组里的值
        for x in nums:
        # 3. 如果x不为0，则把nums[stack]的值替换为x，从0开始，变相的创建了一个新数组
            if x:
                nums[stack] = x
                stack += 1  #替换进去后再加1，这样可以实现按原来的顺序重新排列
            else:
                continue
        for i in range(stack,len(nums)):#数组[stack,n-1]就是为0的元素了
            nums[i] = 0                 #for循环逐一赋值
```
